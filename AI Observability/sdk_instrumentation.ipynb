```python
import os
import sys
from dotenv import load_dotenv

load_dotenv("../.env")   # d'abord, avant de créer les clients

from langfuse.openai import openai
from langfuse import observe, Langfuse

client = openai.OpenAI(
    api_key=os.getenv("GOOGLE_AI_API"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
)

sys.path.append(os.path.abspath(".."))
from shared.shopbot_config import SHOPBOT_SYSTEM_PROMPT
```

```python
langfuse = Langfuse()
LANGFUSE_BASE_URL = os.getenv("LANGFUSE_BASE_URL", "https://cloud.langfuse.com")

MODEL = "gemini-3.5-flash-lite"

print("Setup complete — Langfuse client initialised")
print(f"Traces will appear at: {LANGFUSE_BASE_URL}")
```

```python
@observe(name="ask-shopbot")
def ask_shopbot(user_message: str) -> str:
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": SHOPBOT_SYSTEM_PROMPT},
            {"role": "user", "content": user_message},
        ],
    )
    return response.choices[0].message.content
```

```python
#interacting with AI system and observing the function
ask_shopbot("What is your return policy?")
```
